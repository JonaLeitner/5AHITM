## 1. Health check

```python
import sys, importlib

print("Python:", sys.version.split()[0], "  (should start with 3.11)")
print("Running from:", sys.executable)
print("Environment OK" if "dsaienv" in sys.executable else "WARNING: wrong kernel! Select 'dsaienv' top right.")
print()

missing = []
for pkg in ["numpy", "pandas", "matplotlib", "seaborn", "sklearn", "mlxtend"]:
    try:
        mod = importlib.import_module(pkg)
        print(f"OK       {pkg:<11} {mod.__version__}")
    except ImportError:
        missing.append(pkg)
        print(f"MISSING  {pkg}")

if missing:
    names = " ".join("scikit-learn" if p == "sklearn" else p for p in missing)
    print(f"\nFix in a terminal:  conda install -n dsaienv -y {names}")
    print("Then: restart the kernel (button 'Restart' at the top) and run again.")
else:
    print("\nAll packages installed.")
```

## 2. A plot with love (numpy + matplotlib)
The famous *heart curve*: two formulas with sin and cos.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 400)
x = 16 * np.sin(t) ** 3
y = 13 * np.cos(t) - 5 * np.cos(2 * t) - 2 * np.cos(3 * t) - np.cos(4 * t)

plt.figure(figsize=(4, 4))
plt.fill(x, y, color="crimson")
plt.title("I love data science")
plt.axis("equal"); plt.axis("off")
plt.show()
```
